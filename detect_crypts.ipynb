```python
def load_cell_graph_from_npz(data: np.lib.npyio.NpzFile) -> nx.Graph:
    """
    Reconstruct a NetworkX graph from edges and node count stored in an NPZ file.
    This matches the new preprocessing format where `edges` and `n_nodes`
    are explicitly saved.
    """
    if "graph_edges" not in data.files or "graph_n_nodes" not in data.files:
        raise KeyError("NPZ file must contain 'edges' and 'n_nodes' to load the cell graph.")

    edges = data["graph_edges"]
    n_nodes = int(data["graph_n_nodes"])

    G = nx.Graph()
    G.add_nodes_from(range(n_nodes))

    if edges.size > 0:
        edges = edges.reshape(-1, 2)
        G.add_edges_from(edges.tolist())

    return G
```

```python

import numpy as np
import plotly.graph_objects as go

import numpy as np
import plotly.graph_objects as go

def plot_cell_graph_3D(centroids, graph, field=None, node_size=4, edge_width=1, colorscale="RdBu_r"):
    """
    Plot 3D cell graph with nodes colored by a continuous field (e.g. HKS),
    with a diverging colormap centered at 0.
    """
    x, y, z = centroids[:, 0], centroids[:, 1], centroids[:, 2]

    # Edges
    edge_x, edge_y, edge_z = [], [], []
    for i, j in graph.edges():
        edge_x += [centroids[i, 0], centroids[j, 0], None]
        edge_y += [centroids[i, 1], centroids[j, 1], None]
        edge_z += [centroids[i, 2], centroids[j, 2], None]

    edge_trace = go.Scatter3d(
        x=edge_x, y=edge_y, z=edge_z,
        mode="lines",
        line=dict(width=edge_width, color="lightgray"),
        hoverinfo="none"
    )

    # Nodes
    if field is None:
        field = np.zeros(len(centroids))

    # Center color scale on 0
    vmax = np.nanmax(np.abs(field))
    vmin, vmax = -vmax, vmax

    node_trace = go.Scatter3d(
        x=x, y=y, z=z,
        mode="markers",
        marker=dict(
            size=node_size,
            color=field,
            colorscale=colorscale,
            # cmin=vmin,
            # cmax=vmax,
            colorbar=dict(
                title="",
                x=1.05,
                len=0.8,
                thickness=12
            ),
            line=dict(width=0)
        ),
        hoverinfo="none"
    )

    fig = go.Figure(data=[edge_trace, node_trace])
    fig.update_layout(
        scene=dict(aspectmode="data"),
        margin=dict(l=0, r=0, t=0, b=0)
    )
    fig.show()


```

```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import math

def plot_hks_graph_subplots(
    centroids, graph, fields_cell, scale_labels=None,
    colorscale="RdBu_r", max_cols=3, node_size=4, edge_width=1
):
    """
    Grid of 3D subplots (max 3 columns). Each subplot = HKS at one time scale
    with its own colorbar positioned within that subplot.
    """
    N, S = fields_cell.shape
    if scale_labels is None:
        scale_labels = [f"t[{i}]" for i in range(S)]

    ncols = min(max_cols, S)
    nrows = math.ceil(S / max_cols)

    # Precompute edges once
    edge_x, edge_y, edge_z = [], [], []
    for i, j in graph.edges():
        edge_x += [centroids[i, 0], centroids[j, 0], None]
        edge_y += [centroids[i, 1], centroids[j, 1], None]
        edge_z += [centroids[i, 2], centroids[j, 2], None]

    fig = make_subplots(
        rows=nrows, cols=ncols,
        specs=[[{'type': 'scene'} for _ in range(ncols)] for _ in range(nrows)],
        subplot_titles=scale_labels
    )

    for k in range(S):
        row = k // ncols + 1
        col = k % ncols + 1

        field = fields_cell[:, k]
        vmax = float(np.nanmax(np.abs(field))) if np.any(np.isfinite(field)) else 1.0
        cmin, cmax = -vmax, vmax  # center at 0 per subplot

        # ---- colorbar position for THIS subplot (normalized figure coords) ----
        # Subplot domain: [ (col-1)/ncols , col/ncols ] × [ 1-row/nrows , 1-(row-1)/nrows ]
        x0 = (col - 1) / ncols
        x1 = col / ncols
        y0 = 1 - row / nrows
        y1 = 1 - (row - 1) / nrows
        cb_x = x1 - 0.02                 # a bit inside the right edge of this subplot
        cb_y = (y0 + y1) / 2.0           # vertically centered in this subplot
        cb_len = 0.65 / nrows            # short bar; scales with number of rows

        # Edges
        fig.add_trace(
            go.Scatter3d(
                x=edge_x, y=edge_y, z=edge_z,
                mode="lines",
                line=dict(width=edge_width, color="lightgray"),
                hoverinfo="none",
                showlegend=False
            ),
            row=row, col=col
        )

        # Nodes
        fig.add_trace(
            go.Scatter3d(
                x=centroids[:, 0], y=centroids[:, 1], z=centroids[:, 2],
                mode="markers",
                marker=dict(
                    size=node_size,
                    color=field,
                    colorscale=colorscale,
                    cmin=cmin, cmax=cmax,
                    colorbar=dict(
                        title="",
                        x=cb_x, y=cb_y,
                        yanchor="middle",
                        len=cb_len,
                        thickness=12
                    ),
                    line=dict(width=0)
                ),
                hoverinfo="none",
                showlegend=False,
                name=scale_labels[k]
            ),
            row=row, col=col
        )

        # Keep aspect ratio
        scene_name = "scene" if (row == 1 and col == 1) else f"scene{k+1}"
        fig.update_layout(**{scene_name: dict(aspectmode='data')})

    fig.update_layout(
        height=500 * nrows,
        width=380 * ncols,
        margin=dict(l=10, r=10, t=40, b=10),
        showlegend=False
    )
    fig.show()



plot_hks_graph_subplots(centroids, cell_graph, (cell_weighted_hks-np.mean(cell_weighted_hks, axis=0))/np.std(cell_weighted_hks, axis=0))
# plot_hks_graph_subplots(centroids, cell_graph, cell_weighted_hks)
```

```python
import numpy as np
import networkx as nx
import plotly.graph_objects as go

def plot_organoid_loops_3d(coords: np.ndarray,
                           G: nx.Graph,
                           z_small: np.ndarray,
                           loops: list[np.ndarray],
                           title: str = "") -> go.Figure:
    """
    Minimal 3D plot:
      - nodes colored by z_small (RdBu_r) with symmetric limits around 0
      - light wireframe edges
      - each loop as a thick polyline in a distinct color
    """
    XYZ = coords[:, :3]

    # wireframe
    ex, ey, ez = [], [], []
    for u, v in G.edges():
        ex += [XYZ[u,0], XYZ[v,0], None]
        ey += [XYZ[u,1], XYZ[v,1], None]
        ez += [XYZ[u,2], XYZ[v,2], None]

    traces = []
    if ex:
        traces.append(go.Scatter3d(
            x=ex, y=ey, z=ez, mode="lines",
            line=dict(width=1.0, color="rgba(120,120,120,0.75)"),
            opacity=1.0, hoverinfo="skip", name="wireframe", showlegend=False
        ))

    # symmetric color limits so 0 is centered
    finite = np.isfinite(z_small)
    vmax = float(np.nanmax(np.abs(z_small[finite]))) if np.any(finite) else 1.0
    if vmax == 0.0:
        vmax = 1.0  # avoid zero range

    # nodes
    traces.append(go.Scatter3d(
        x=XYZ[:,0], y=XYZ[:,1], z=XYZ[:,2],
        mode="markers",
        marker=dict(
            size=3.5,
            color=z_small,
            colorscale="RdBu_r",
            showscale=True,
            colorbar=dict(title="z-HKS (small t)"),
            opacity=0.95,
            cmin=-vmax, cmax=vmax, cmid=0.0
        ),
        hoverinfo="skip", name="cells", showlegend=False
    ))

    # loops
    palette = ["#e41a1c", "#377eb8", "#4daf4a", "#984ea3",
               "#ff7f00", "#a65628", "#f781bf", "#999999"]
    for i, poly in enumerate(loops):
        if poly.size < 2:
            continue
        # ensure closed
        if poly[0] != poly[-1]:
            poly = np.concatenate([poly, poly[:1]])
        traces.append(go.Scatter3d(
            x=XYZ[poly,0], y=XYZ[poly,1], z=XYZ[poly,2],
            mode="lines",
            line=dict(width=7.0, color=palette[i % len(palette)]),
            hoverinfo="skip", name=f"loop {i+1}", showlegend=True
        ))

    mins = XYZ.min(0); maxs = XYZ.max(0); span = float(np.max(maxs - mins)); ctr = (mins + maxs)/2
    fig = go.Figure(traces)
    fig.update_layout(
        scene=dict(
            xaxis=dict(range=[ctr[0]-span/2, ctr[0]+span/2], showgrid=False, showticklabels=False, zeroline=False),
            yaxis=dict(range=[ctr[1]-span/2, ctr[1]+span/2], showgrid=False, showticklabels=False, zeroline=False),
            zaxis=dict(range=[ctr[2]-span/2, ctr[2]+span/2], showgrid=False, showticklabels=False, zeroline=False),
            aspectmode="cube",
        ),
        margin=dict(l=0,r=0,t=30,b=0),
        title=title,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1.0)
    )
    return fig



import numpy as np
import networkx as nx

def _euclid(a, b, XYZ):
    pa, pb = XYZ[a], XYZ[b]
    return float(np.linalg.norm(pa - pb))

def _smooth_cost(a, b, c, XYZ):
    # squared second-difference in R^3 (curvature proxy)
    pa, pb, pc = XYZ[a], XYZ[b], XYZ[c]
    return float(np.linalg.norm(pc - 2*pb + pa)**2)

def _ring_dp_min_energy_cycle(G, XYZ, z_small, seed_loop, *,
                              r_hops=2,                # search radius around each seed node
                              K_max=20,                # cap #candidates per position to keep DP tractable
                              w_data=1.0, w_len=0.15, w_smooth=0.5,
                              prefer_neck_mask=None,   # optional boolean mask to bias neck nodes
                              neck_bonus=0.3):         # subtract from z_small if node is neck (helps “snap” in)
    """
    Dynamic programming on a ring (closed contour) with local candidate sets per position.
    Returns (best_cycle_nodes as np.ndarray) or None if no feasible cycle found.
    """
    m = int(len(seed_loop))
    if m < 3: return None

    # 1) Build candidate sets C[i] near each seed vertex
    #    Use BFS up to r_hops, score by z_small (asc) + distance to seed, keep top K_max
    C = []
    for i, v in enumerate(seed_loop):
        # BFS
        fringe = {int(v)}
        visited = {int(v)}
        for _ in range(r_hops):
            newf = set()
            for u in fringe:
                for w in G.neighbors(u):
                    if w not in visited:
                        visited.add(int(w))
                        newf.add(int(w))
            fringe = newf

        cand = list(visited)
        # score candidates: low z_small and near the seed
        zs = z_small[cand].astype(float)
        if prefer_neck_mask is not None:
            # give neck nodes a little extra “pull”
            zs = zs - neck_bonus * prefer_neck_mask[np.array(cand, int)]
        d0 = np.array([_euclid(v, u, XYZ) for u in cand], dtype=float)
        score = 1.0*zs + 0.05*d0
        order = np.argsort(score)
        sel = [cand[j] for j in order[:min(K_max, len(order))]]
        C.append(sel)

    # Early out if any C[i] empty
    if any(len(ci) == 0 for ci in C):
        return None

    # 2) Ring DP: we unroll the ring by fixing a start candidate c0 in C[0],
    #    then solve a linear-chain DP for i=1..m-1 with a 2nd-order smooth term.
    #    Finally, we add the (m-1)->0 closure costs and pick the best start.
    best_total = np.inf
    best_seq = None

    # Precompute pairwise euclid between all candidates of adjacent positions to speed up DP
    # and also precompute data term per candidate
    data_costs = [np.array([float(z_small[u]) for u in Ci], dtype=float) for Ci in C]
    if prefer_neck_mask is not None:
        for i, Ci in enumerate(C):
            data_costs[i] = data_costs[i] - neck_bonus * prefer_neck_mask[np.array(Ci, int)]

    # adjacency feasibility (must be connected in G between consecutive picks)
    def feasible(u, v):
        return G.has_edge(int(u), int(v)) or (u == v)

    for s_idx, s in enumerate(C[0]):
        # DP states keep (prev_index, energy) per candidate at position i
        # We need a 2nd-order term, so we carry pairs. Implementation trick:
        # - forward pass keeps best parent per (i, j) referencing candidate index at i-1.
        parents = []
        E_prev = None  # energies for position 0..i
        idx_prev = None

        # Initialize at i=0 with fixed start s_idx
        E0 = np.full((len(C[0]),), np.inf, dtype=float)
        E0[s_idx] = w_data * data_costs[0][s_idx]
        parents.append([(-1, -1)]*len(C[0]))
        E_prev = E0
        idx_prev = np.arange(len(C[0]))

        # i=1: only length + data (no smooth term yet)
        # Build E1[j] = E0[s_idx] + w_len*|x1 - x0| + w_data*data1
        E1 = np.full((len(C[1]),), np.inf, dtype=float)
        par1 = [(-1, -1)]*len(C[1])
        for j, v1 in enumerate(C[1]):
            if not feasible(C[0][s_idx], v1): 
                continue
            e = E_prev[s_idx] + w_len * _euclid(C[0][s_idx], v1, XYZ) + w_data * data_costs[1][j]
            E1[j] = e
            par1[j] = (0, s_idx)
        parents.append(par1)
        E_prev = E1

        # i=2..m-1: use full energy with smoothness on triples
        for i in range(2, m):
            Ei = np.full((len(C[i]),), np.inf, dtype=float)
            pari = [(-1, -1)]*len(C[i])
            for j, vj in enumerate(C[i]):
                # try all parents (k at i-1, p at i-2)
                best_e = np.inf; best_k = -1
                for k, vk in enumerate(C[i-1]):
                    if not feasible(vk, vj): 
                        continue
                    e_len = w_len * _euclid(vk, vj, XYZ)
                    # we need i-2 choices: look into parents[i-1][k] to know its parent index at i-2
                    pk, idx_pk = parents[i-1][k]
                    if pk == -1: 
                        continue
                    vp = C[i-2][idx_pk]
                    e_smooth = w_smooth * _smooth_cost(vp, vk, vj, XYZ)
                    e = E_prev[k] + e_len + e_smooth + w_data * data_costs[i][j]
                    if e < best_e:
                        best_e = e; best_k = k
                if best_k != -1:
                    Ei[j] = best_e
                    pari[j] = (i-1, best_k)
            parents.append(pari)
            E_prev = Ei

        # Close the ring: add smooth+length costs for linking (m-1)→0 and (m-2, m-1, 0) triple
        # among candidates that connect to s.
        last_i = m-1
        best_e_total = np.inf
        best_last = -1
        for j, v_last in enumerate(C[last_i]):
            if not feasible(v_last, s):
                continue
            pk, idx_pk = parents[last_i][j]
            if pk == -1: 
                continue
            v_prev = C[last_i-1][idx_pk]
            e_len_close = w_len * _euclid(v_last, s, XYZ)
            e_smooth_close = w_smooth * _smooth_cost(v_prev, v_last, s, XYZ)
            e_total = E_prev[j] + e_len_close + e_smooth_close
            if e_total < best_e_total:
                best_e_total = e_total
                best_last = j

        if best_last == -1:
            continue  # no feasible closure for this start

        if best_e_total < best_total:
            # backtrack indices
            seq_idx = [None]*m
            seq_idx[-1] = best_last
            for i in range(m-1, 0, -1):
                pi, pj = parents[i][seq_idx[i]]
                seq_idx[i-1] = pj
            seq_nodes = np.array([C[i][seq_idx[i]] for i in range(m)], dtype=int)
            best_total = best_e_total
            best_seq = seq_nodes

    return best_seq  # may be None


def _simple_cycle_from_selected(G, nodes_ordered):
    """
    Given an ordered selection (possibly with repeats), extract a simple cycle:
    - induce a subgraph on selected nodes + their 1-hop neighbors
    - take cycle_basis and pick the one with max overlap and largest length
    """
    sel = set(int(x) for x in nodes_ordered)
    cand = set(sel)
    for u in list(sel):
        cand.update(int(v) for v in G.neighbors(u))
    H = G.subgraph(list(cand)).copy()
    if H.number_of_nodes() < 3: 
        return None
    cycles = nx.cycle_basis(H)
    if not cycles: 
        return None
    Lset = sel
    # prefer cycles with max overlap, then longer ones
    best = max(cycles, key=lambda c: (len(set(c) & Lset), len(c)))
    cyc = np.array(best, dtype=int)
    # ensure it’s closed (your plotter adds closure anyway)
    return cyc


def refine_loops_with_graph_snake(G, coords, z_small, seed_loops,
                                  *,
                                  r_hops=2, K_max=20,
                                  w_data=1.0, w_len=0.15, w_smooth=0.5,
                                  neck_mask=None,
                                  min_crypt_cells=5,
                                  crypt_pos_frac=0.7):
    """
    Tighten/smooth each seed loop via a discrete active contour (ring DP),
    then enforce separation + crypt constraints. Returns a list of refined loops.
    Safe to plug between your Step 4 and Step 6.
    """
    XYZ = coords[:, :3]
    refined = []

    def separates(loop_poly):
        # identical to your _touching_components_for_loop logic but simplified here
        loop_set = set(int(x) for x in loop_poly)
        G_cut = G.copy(); G_cut.remove_nodes_from(loop_set)
        touching = []
        for cc in nx.connected_components(G_cut):
            cc = np.array(sorted(list(cc)), dtype=int)
            hit = False
            for n in cc:
                for nbr in G.neighbors(int(n)):
                    if nbr in loop_set:
                        hit = True; break
                if hit: break
            if hit: touching.append(cc)
        return (len(touching) >= 2), touching

    def crypt_positive(crypt_nodes):
        if crypt_nodes.size == 0: return False
        pos_all = np.all(z_small[crypt_nodes, None] > 0, axis=1)  # single-scale here; expand if multi-scale
        return (float(np.mean(pos_all)) >= crypt_pos_frac)

    for lp in seed_loops:
        seed = np.array(lp, dtype=int)
        cand = _ring_dp_min_energy_cycle(
            G, XYZ, z_small, seed,
            r_hops=r_hops, K_max=K_max,
            w_data=w_data, w_len=w_len, w_smooth=w_smooth,
            prefer_neck_mask=(neck_mask.astype(float) if neck_mask is not None else None),
            neck_bonus=0.3
        )
        if cand is None:
            # fallback: try just projecting the seed to a simple cycle
            simp = _simple_cycle_from_selected(G, seed)
            if simp is None: 
                continue
            cand = simp
        else:
            simp = _simple_cycle_from_selected(G, cand)
            if simp is not None:
                cand = simp

        ok, touching = separates(cand)
        if not ok:
            # if we lost separation, gently expand by one hop and re-project once
            band = set(int(x) for x in cand)
            for u in list(band):
                band.update(int(v) for v in G.neighbors(u))
            H = G.subgraph(list(band)).copy()
            cyc = nx.cycle_basis(H)
            if cyc:
                # pick the cycle with max overlap to cand
                Lset = set(int(x) for x in cand)
                best = max(cyc, key=lambda c: (len(set(c)&Lset), len(c)))
                cand = np.array(best, dtype=int)
                ok, touching = separates(cand)
        if not ok:
            continue

        touching.sort(key=lambda a: a.size)
        crypt = touching[0]
        if crypt.size < min_crypt_cells:
            continue
        if not crypt_positive(crypt):
            # small, local “balloon” step: try one more DP pass with a tiny outward bias
            cand2 = _ring_dp_min_energy_cycle(
                G, XYZ, z_small - 0.1, cand,  # nudge data term to prefer even-lower HKS
                r_hops=r_hops, K_max=K_max,
                w_data=w_data, w_len=w_len, w_smooth=w_smooth,
                prefer_neck_mask=(neck_mask.astype(float) if neck_mask is not None else None),
                neck_bonus=0.4
            )
            if cand2 is not None:
                simp2 = _simple_cycle_from_selected(G, cand2)
                if simp2 is not None: cand2 = simp2
                ok2, touching2 = separates(cand2)
                if ok2:
                    touching2.sort(key=lambda a: a.size)
                    crypt2 = touching2[0]
                    if crypt2.size >= min_crypt_cells and crypt_positive(crypt2):
                        cand = cand2
        refined.append(cand)

    return refined



import numpy as np
import networkx as nx

def dissect_by_necks(C: np.ndarray,
                     G: nx.Graph,
                     coords: np.ndarray,
                     *,
                     small_neg_thr: float = -0.5,   # neck: z_small <= this
                     large_near0_thr: float = 0.35, # neck: z_large <= this
                     min_loop_nodes: int = 8,       # discard tiny rings
                     band_hops: int = 1,            # neck band thinness (<= 1 hop off cycle)
                     min_crypt_cells: int = 5,      # interior region size guard
                     crypt_pos_frac: float = 0.7,   # interior positivity across all scales
                     do_plots: bool = True
                     ) -> list[nx.Graph]:
    """
    Clean neck-based dissection with minimal knobs.
    Steps:
      1) Neck candidates: small<=thr & |large|<=thr
      2) Extract loops per component (MCB) + split figure-eight (<=2 subloops)
      3) Thinness: keep loops whose band thickness <= band_hops
      4) Size: drop loops with < min_loop_nodes nodes
      5) Interior sanity: loop encloses crypt-like interior (size + positivity)
      6) Deduplicate: drop near-duplicate loops (Jaccard > 0.8, keep longer)
      7) Cut graph along union of loop nodes → return connected components
    """
    Z = C  # already z-scored
    z_small = Z[:, 0]
    z_large = Z[:, -1]

    def _print_and_plot(stage: str, loops: list[np.ndarray]):
        print(f"[{stage}] loops: {len(loops)}")
        if do_plots:
            fig = plot_organoid_loops_3d(coords, G, z_small, loops, title=f"{stage} — {len(loops)} loops")
            fig.show()

    # ------------ Step 1: neck candidates ------------
    neck_mask = (z_small <= small_neg_thr) & (z_large <= large_near0_thr)
    neck_sub = G.subgraph(np.where(neck_mask)[0]).copy()
    cand_components = [np.array(sorted(list(cc)), dtype=int) for cc in nx.connected_components(neck_sub)]
    # for visualization, show largest cycle from each component as draft loops
    draft_loops = []
    for cc in cand_components:
        H = neck_sub.subgraph(cc).copy()
        cycles = nx.cycle_basis(nx.k_core(H, 2)) if H.number_of_nodes() >= 3 else []
        if cycles:
            cyc = max(cycles, key=len)
            draft_loops.append(np.array(list(cyc), dtype=int))
    _print_and_plot("Step 1: candidates (draft cycles)", draft_loops)

    # ------------ Step 2: extract loops (MCB) & split figure-eight ------------
    def _thinness_ok_local(H_int: nx.Graph, loop_nodes_int: np.ndarray,
                        band_hops: int, max_outlier_frac: float = 0.05) -> bool:
        """Thinness measured only on the band around the loop (local),
        allowing a small outlier fraction."""
        from collections import deque
        loop_set = set(int(x) for x in loop_nodes_int)

        # BFS distances from the loop
        dist = {n: np.inf for n in H_int.nodes()}
        dq = deque()
        for n in loop_set:
            dist[n] = 0
            dq.append(n)
        while dq:
            u = dq.popleft()
            if dist[u] >= band_hops:
                continue
            for v in H_int.neighbors(u):
                if dist[v] > dist[u] + 1:
                    dist[v] = dist[u] + 1
                    dq.append(v)

        # Only consider nodes within (band_hops+1) of the loop as the "band"
        band_nodes = [n for n, d in dist.items() if d <= band_hops]
        if not band_nodes:
            return False

        # Now verify that almost all band nodes are actually close (<= band_hops)
        # (This is trivially true by construction, but we keep the interface consistent
        #  and allow a tiny fraction of missing neighbors if the loop is fragmented)
        outliers = [n for n in band_nodes if dist[n] > band_hops]
        return (len(outliers) / max(1, len(band_nodes))) <= max_outlier_frac

    def _extract_loops_from_component(H: nx.Graph) -> list[np.ndarray]:
        # relabel to ints for MCB robustness
        nodes_orig = list(H.nodes())
        to_int = {n:i for i,n in enumerate(nodes_orig)}
        to_orig = {i:n for n,i in to_int.items()}
        H_int = nx.relabel_nodes(H, to_int, copy=True)
        mcb = nx.minimum_cycle_basis(H_int)
        if not mcb: return []
        mcb_sorted = sorted(mcb, key=len, reverse=True)

        def to_orig_cycle(c_int: list[int]) -> np.ndarray:
            return np.array([to_orig[int(u)] for u in c_int], dtype=int)

        loops = []
        main = np.array(mcb_sorted[0], dtype=int)
        if _thinness_ok_local(H_int, main, band_hops):
            loops.append(to_orig_cycle(main))
        # try a second, distinct loop if cyclomatic number >= 2
        if (H_int.number_of_edges() - H_int.number_of_nodes() + 1) >= 2 and len(mcb_sorted) > 1:
            a = set(main.tolist())
            for c in mcb_sorted[1:]:
                b = set(np.array(c, dtype=int).tolist())
                jacc = len(a & b) / max(1, len(a | b))
                if jacc > 0.8:  # too similar to main
                    continue
                if _thinness_ok_local(H_int, np.array(c, dtype=int), band_hops):
                    loops.append(to_orig_cycle(c))
                    break
        return loops[:2]  # at most two

    loops_step2: list[np.ndarray] = []
    for cc in cand_components:
        Hc = neck_sub.subgraph(cc).copy()
        if Hc.number_of_nodes() < 3:
            continue

        # Split into biconnected blocks; run the extractor on each block
        for block_nodes in nx.biconnected_components(Hc):
            B = Hc.subgraph(block_nodes).copy()
            if B.number_of_edges() - B.number_of_nodes() + 1 <= 0:
                continue  # no cycles in this block
            loops_step2.extend(_extract_loops_from_component(B))
    _print_and_plot("Step 2: extracted loops (+ figure-eight split)", loops_step2)

    # ------------ Step 3: thinness already enforced in extraction ------------
    # (kept as a step for clarity; nothing else to do here)
    loops_step3 = loops_step2
    _print_and_plot("Step 3: thinness check (band hops)", loops_step3)

    # ------------ Step 4: drop small loops ------------
    loops_step4 = [lp for lp in loops_step3 if lp.size >= min_loop_nodes]
    _print_and_plot("Step 4: size filter", loops_step4)


    # ------------ Step 5: ensure separation, assign neck to villus, shrink-by-lowest z_small (PRESERVE ORDER) ------------
    def _touching_components_for_loop(loop_poly: np.ndarray) -> list[np.ndarray]:
        """Cut only this loop; return components that touch the loop, sorted by size (asc)."""
        loop_set = set(int(x) for x in loop_poly)
        G_cut = G.copy()
        G_cut.remove_nodes_from(loop_set)
        touching: list[np.ndarray] = []
        for cc in nx.connected_components(G_cut):
            cc_nodes = np.array(sorted(list(cc)), dtype=int)
            # touches loop if any node has neighbor in loop
            hit = False
            for n in cc_nodes:
                for nbr in G.neighbors(int(n)):
                    if nbr in loop_set:
                        hit = True
                        break
                if hit: break
            if hit:
                touching.append(cc_nodes)
        touching.sort(key=lambda a: a.size)  # smallest first (crypt), largest last (villus)
        return touching

    def _crypt_is_positive(crypt_nodes: np.ndarray) -> bool:
        pos_all = np.all(Z[crypt_nodes, :] > 0, axis=1)
        return (float(np.mean(pos_all)) >= crypt_pos_frac)

    def _order_loop_nodes(G: nx.Graph, loop_nodes: np.ndarray) -> np.ndarray:
        """Return an ordered simple cycle from the induced subgraph on loop_nodes."""
        H = G.subgraph([int(x) for x in loop_nodes]).copy()
        if H.number_of_nodes() < 3:
            return np.array(loop_nodes, dtype=int)
        cycles = nx.cycle_basis(H)
        if not cycles:
            return np.array(loop_nodes, dtype=int)
        Lset = set(int(x) for x in loop_nodes)
        best = max(cycles, key=lambda c: (len(set(c) & Lset), len(c)))
        return np.array(best, dtype=int)

    def _ordered_cycle_including_node(G: nx.Graph,
                                    neck_mask: np.ndarray,
                                    loop_poly: np.ndarray,
                                    include_node: int) -> np.ndarray:
        """
        Build an ordered cycle that includes `include_node`, preferring
        (1) cycles with maximal overlap with current loop, and
        (2) longer cycles (so we don't shrink to triangles).
        """
        cand = set(int(x) for x in loop_poly)
        for u in list(cand):
            for v in G.neighbors(u):
                if neck_mask[v]:
                    cand.add(int(v))
        cand.add(int(include_node))

        Hc = G.subgraph(list(cand)).copy()
        if Hc.number_of_nodes() < 3:
            return loop_poly

        cycles = nx.cycle_basis(Hc)
        if not cycles:
            return loop_poly

        Lset = set(int(x) for x in loop_poly)
        best = None
        best_key = (-1, -1)  # (overlap_count, length)
        for c in cycles:
            if int(include_node) not in c:
                continue
            cset = set(int(x) for x in c)
            key = (len(cset & Lset), len(c))
            if key > best_key:
                best_key = key
                best = c

        if best is None:
            return loop_poly
        return np.array(best, dtype=int)


    def _separates(G: nx.Graph, loop_poly: np.ndarray) -> bool:
        """True if removing loop nodes yields at least 2 touching components."""
        return (len(_touching_components_for_loop(loop_poly)) >= 2)

    # We'll need neck candidates here for sealing; recompute quickly with current thresholds.
    neck_mask_for_step5 = (z_small <= small_neg_thr) & (z_large <= large_near0_thr)

    loops_step5: list[np.ndarray] = []

    for lp_idx, lp in enumerate(loops_step4):
        loop = np.array(lp, dtype=int)  # keep ORDERED polyline

        # --- 5A) Ensure the loop is a true separator: try to seal it (prefer neck, allow fallback) ---
        if not _separates(G, loop):
            print(f"[Step 5] loop#{lp_idx+1}: not separating; attempting to seal the loop (prefer neck, allow fallback)...")
            max_seal_iters = 200
            it = 0
            tried = set()  # nodes we already attempted
            while it < max_seal_iters and not _separates(G, loop):
                it += 1

                # --- DEBUG: print + plot current loop before trying to add anything ---
                print(f"[Step 5][seal] loop#{lp_idx+1} it={it}: pre-add | loop len={loop.size}")
                if do_plots:
                    fig = plot_organoid_loops_3d(coords, G, z_large, [loop],
                                                title=f"Step 5 • seal pre (loop#{lp_idx+1}, it={it})")
                    fig.show()

                loop_set = set(int(x) for x in loop)

                # 1-hop candidates: neck first, then non-neck
                cand1_neck, cand1_fall = [], []
                for u in loop_set:
                    for v in G.neighbors(u):
                        if v in loop_set or v in tried:
                            continue
                        (cand1_neck if neck_mask_for_step5[v] else cand1_fall).append(int(v))

                # 2-hop fallback if no 1-hop candidates
                cand2_neck, cand2_fall = [], []
                if not cand1_neck and not cand1_fall:
                    for u in loop_set:
                        for v in G.neighbors(u):
                            if v in loop_set:
                                continue
                            for w in G.neighbors(v):
                                if w in loop_set or w in tried:
                                    continue
                                (cand2_neck if neck_mask_for_step5[w] else cand2_fall).append(int(w))

                # Merge candidate lists with priority and z_small sort (ascending)
                cand = []
                for group in (cand1_neck, cand1_fall, cand2_neck, cand2_fall):
                    if group:
                        group = np.array(sorted(set(group)), dtype=int)
                        group = group[np.argsort(z_small[group])]
                        cand.extend(group.tolist())

                print(f"[Step 5][seal] loop#{lp_idx+1} it={it}: candidates: "
                    f"1hop-neck={len(cand1_neck)}, 1hop-fallback={len(cand1_fall)}, "
                    f"2hop-neck={len(cand2_neck)}, 2hop-fallback={len(cand2_fall)}")

                if not cand:
                    print(f"[Step 5][seal] loop#{lp_idx+1} it={it}: no candidates left (even with fallback).")
                    break

                progressed = False
                for add_node in cand:
                    tried.add(add_node)

                    new_loop = _ordered_cycle_including_node(G, neck_mask_for_step5, loop, add_node)

                    # Reject if no change
                    if set(new_loop.tolist()) == set(loop.tolist()):
                        continue

                    # Guard 1: don't shrink too much (keep ≥80% of original length)
                    if new_loop.size < int(0.8 * loop.size):
                        # Tiny triangles etc. — skip this candidate
                        print(f"[Step 5][seal] loop#{lp_idx+1} it={it}: REJECT new cycle len={new_loop.size} "
                            f"(shrinks from {loop.size}).")
                        continue

                    # Guard 2: must reuse enough of the original loop nodes (≥50% overlap)
                    overlap = len(set(new_loop.tolist()) & set(loop.tolist()))
                    if overlap < int(0.5 * loop.size):
                        print(f"[Step 5][seal] loop#{lp_idx+1} it={it}: REJECT new cycle overlap={overlap} "
                            f"(old len={loop.size}).")
                        continue

                    # Accept this candidate
                    loop = new_loop
                    print(f"[Step 5][seal] loop#{lp_idx+1} it={it}: added node {add_node} "
                        f"(neck={bool(neck_mask_for_step5[add_node])}, z_small={z_small[add_node]:.3f}); "
                        f"new loop len={loop.size}")




                    print(f"[Step 5][seal] loop#{lp_idx+1} it={it}: added node {add_node} "
                        f"(neck={bool(neck_mask_for_step5[add_node])}, z_small={z_small[add_node]:.3f}); "
                        f"new loop len={loop.size}")

                    # --- DEBUG: plot after adding node ---
                    if do_plots:
                        fig = plot_organoid_loops_3d(coords, G, z_large, [loop],
                                                    title=f"Step 5 • seal post-add {add_node} "
                                                        f"(loop#{lp_idx+1}, it={it})")
                        fig.show()

                    progressed = True
                    if _separates(G, loop):
                        print(f"[Step 5][seal] loop#{lp_idx+1} it={it}: loop now separates.")
                        break
                    # else continue to try more adds within this iteration

                if not progressed:
                    print(f"[Step 5][seal] loop#{lp_idx+1} it={it}: made no progress (cycle couldn’t absorb candidates).")
                    break

            if not _separates(G, loop):
                # Optional quick leak note
                print(f"[Step 5] DROP loop#{lp_idx+1}: could not seal to a separating cycle (even with fallback).")
                continue
            else:
                print(f"[Step 5] loop#{lp_idx+1}: sealed successfully.")


        # --- 5B) Assign neck to villus; smallest adjacent piece is crypt candidate ---
        touching = _touching_components_for_loop(loop)
        if len(touching) == 0:
            print(f"[Step 5] DROP loop#{lp_idx+1}: no adjacent components after cut.")
            continue
        crypt = touching[0]
        villus = touching[-1]
        print(f"[Step 5] loop#{lp_idx+1}: initial crypt size={crypt.size}, villus size={villus.size}, loop len={loop.size}")

        # size guards
        if crypt.size < min_crypt_cells:
            print(f"[Step 5] DROP loop#{lp_idx+1}: crypt too small ({crypt.size} < {min_crypt_cells}).")
            continue
        if loop.size < min_loop_nodes:
            print(f"[Step 5] DROP loop#{lp_idx+1}: loop too small ({loop.size} < {min_loop_nodes}).")
            continue

        # If crypt is positive, keep
        if _crypt_is_positive(crypt):
            print(f"[Step 5] KEEP loop#{lp_idx+1}: crypt positive without shrink.")
            loops_step5.append(loop)
            continue

        # --- 5C) Shrink crypt iteratively by moving lowest z_small boundary crypt cell into the loop ---
        max_iters = 200
        it = 0
        rejected_reason = None
        while True:
            it += 1
            if it > max_iters:
                rejected_reason = f"max iters exceeded ({max_iters}) while shrinking"
                crypt = np.array([], dtype=int)
                break

            # boundary crypt nodes (crypt nodes adjacent to loop)
            loop_set = set(int(x) for x in loop)
            boundary = []
            for n in crypt:
                if any((nbr in loop_set) for nbr in G.neighbors(int(n))):
                    boundary.append(int(n))
            if not boundary:
                rejected_reason = "no boundary crypt nodes left to shrink"
                crypt = np.array([], dtype=int)
                break

            # pick boundary node with smallest z_small
            b = np.array(boundary, dtype=int)
            move = int(b[np.argmin(z_small[b])])
            print(f"[Step 5] loop#{lp_idx+1} it={it}: move boundary node {move} (z_small={z_small[move]:.3f}) into loop")

            # rebuild a cycle that includes this node (within neck band) to PRESERVE ORDER and contiguity
            loop = _ordered_cycle_including_node(G, neck_mask_for_step5, loop, move)

            # after changing the loop, ensure it still separates
            if not _separates(G, loop):
                rejected_reason = "loop lost separation after shrink step"
                crypt = np.array([], dtype=int)
                break

            # re-cut and re-identify smallest/largest touching comps
            touching = _touching_components_for_loop(loop)
            if len(touching) == 0:
                rejected_reason = "degenerate cut after shrink (no touching components)"
                crypt = np.array([], dtype=int)
                break
            crypt = touching[0]
            villus = touching[-1]
            print(f"[Step 5] loop#{lp_idx+1} it={it}: new crypt size={crypt.size}, loop len={loop.size}")

            # size guards after update
            if crypt.size < min_crypt_cells:
                rejected_reason = f"crypt too small after shrink ({crypt.size} < {min_crypt_cells})"
                crypt = np.array([], dtype=int)
                break
            if loop.size < min_loop_nodes:
                rejected_reason = f"loop too small after shrink ({loop.size} < {min_loop_nodes})"
                crypt = np.array([], dtype=int)
                break

            # positivity?
            if _crypt_is_positive(crypt):
                print(f"[Step 5] KEEP loop#{lp_idx+1}: crypt positive after shrink (iters={it}, crypt size={crypt.size}).")
                break

        if crypt.size == 0:
            print(f"[Step 5] DROP loop#{lp_idx+1}: {rejected_reason}.")
            continue

        loops_step5.append(loop)

    _print_and_plot("Step 5: separation+assign-to-villus+shrink (order preserved, sealing if needed)", loops_step5)



    # Compute a neck mask consistent with your thresholds (so the snake snaps to necks)
    neck_mask = (Z[:,0] <= small_neg_thr) & (Z[:,-1] <= large_near0_thr)

    # Refine step-6 loops
    loops_step5 = refine_loops_with_graph_snake(
        cell_graph, centroids, Z[:,0],  # use your “small t” HKS as curvature proxy
        loops_step5,
        r_hops=1, K_max=20,
        w_data=1.0, w_len=0.15, w_smooth=2.0,   # good starting weights
        neck_mask=neck_mask,
        min_crypt_cells=min_crypt_cells,
        crypt_pos_frac=crypt_pos_frac
    )

    _print_and_plot("Step 5.5: loop optimization", loops_step5)


    # ------------ Step 6: deduplicate near-identical loops ------------
    def _dedup(loops: list[np.ndarray]) -> list[np.ndarray]:
        keep: list[np.ndarray] = []
        for lp in sorted(loops, key=lambda a: -a.size):
            aset = set(lp.tolist())
            dup = False
            for kp in keep:
                bset = set(kp.tolist())
                jacc = len(aset & bset) / max(1, len(aset | bset))
                if jacc > 0.8:
                    dup = True; break
            if not dup:
                keep.append(lp)
        return keep
    loops_final = _dedup(loops_step5)
    _print_and_plot("Step 6: deduplicate", loops_final)

    # ------------ Step 7: cut graph along loop nodes ------------
    if len(loops_final) == 0:
        _print_and_plot("Step 7: no loops — return whole graph", [])
        return [G.copy()]
    cut_nodes = np.unique(np.concatenate(loops_final))
    G_cut = G.copy(); G_cut.remove_nodes_from(cut_nodes)
    pieces = [G.subgraph(cc).copy() for cc in nx.connected_components(G_cut)]
    _print_and_plot("Step 7: cut & components", loops_final)
    return pieces



HKS_zscored = (cell_weighted_hks-np.mean(cell_weighted_hks, axis=0))/np.std(cell_weighted_hks, axis=0)

components = dissect_by_necks(
    HKS_zscored[:,:7], cell_graph, centroids,
    small_neg_thr= -0.1,
    large_near0_thr=0.4,
    min_loop_nodes=8,
    band_hops=2,
    min_crypt_cells=5,
    crypt_pos_frac=0.5,
    do_plots=True  # set False to turn off plotting/printing
)
print("num components:", len(components))
```

```python
import numpy as np
import networkx as nx
import plotly.graph_objects as go

def plot_organoid_loops_3d(coords: np.ndarray,
                           G: nx.Graph,
                           z_small: np.ndarray,
                           loops: list[np.ndarray],
                           title: str = "") -> go.Figure:
    """
    Minimal 3D plot:
      - nodes colored by z_small (RdBu_r) with symmetric limits around 0
      - light wireframe edges
      - each loop as a thick polyline in a distinct color
    """
    XYZ = coords[:, :3]

    # wireframe
    ex, ey, ez = [], [], []
    for u, v in G.edges():
        ex += [XYZ[u,0], XYZ[v,0], None]
        ey += [XYZ[u,1], XYZ[v,1], None]
        ez += [XYZ[u,2], XYZ[v,2], None]

    traces = []
    if ex:
        traces.append(go.Scatter3d(
            x=ex, y=ey, z=ez, mode="lines",
            line=dict(width=1.0, color="rgba(120,120,120,0.75)"),
            opacity=1.0, hoverinfo="skip", name="wireframe", showlegend=False
        ))

    # symmetric color limits so 0 is centered
    finite = np.isfinite(z_small)
    vmax = float(np.nanmax(np.abs(z_small[finite]))) if np.any(finite) else 1.0
    if vmax == 0.0:
        vmax = 1.0  # avoid zero range

    # nodes
    traces.append(go.Scatter3d(
        x=XYZ[:,0], y=XYZ[:,1], z=XYZ[:,2],
        mode="markers",
        marker=dict(
            size=3.5,
            color=z_small,
            colorscale="RdBu_r",
            showscale=True,
            colorbar=dict(title="z-HKS (small t)"),
            opacity=0.95,
            cmin=-vmax, cmax=vmax, cmid=0.0
        ),
        hoverinfo="skip", name="cells", showlegend=False
    ))

    # loops
    palette = ["#e41a1c", "#377eb8", "#4daf4a", "#984ea3",
               "#ff7f00", "#a65628", "#f781bf", "#999999"]
    for i, poly in enumerate(loops):
        if poly.size < 2:
            continue
        # ensure closed
        if poly[0] != poly[-1]:
            poly = np.concatenate([poly, poly[:1]])
        traces.append(go.Scatter3d(
            x=XYZ[poly,0], y=XYZ[poly,1], z=XYZ[poly,2],
            mode="lines",
            line=dict(width=7.0, color=palette[i % len(palette)]),
            hoverinfo="skip", name=f"loop {i+1}", showlegend=True
        ))

    mins = XYZ.min(0); maxs = XYZ.max(0); span = float(np.max(maxs - mins)); ctr = (mins + maxs)/2
    fig = go.Figure(traces)
    fig.update_layout(
        scene=dict(
            xaxis=dict(range=[ctr[0]-span/2, ctr[0]+span/2], showgrid=False, showticklabels=False, zeroline=False),
            yaxis=dict(range=[ctr[1]-span/2, ctr[1]+span/2], showgrid=False, showticklabels=False, zeroline=False),
            zaxis=dict(range=[ctr[2]-span/2, ctr[2]+span/2], showgrid=False, showticklabels=False, zeroline=False),
            aspectmode="cube",
        ),
        margin=dict(l=0,r=0,t=30,b=0),
        title=title,
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1.0)
    )
    return fig


# --------------------------- OPTIMIZER CORE ---------------------------

def _euclid(a, b, XYZ):
    pa, pb = XYZ[a], XYZ[b]
    return float(np.linalg.norm(pa - pb))

def _smooth_cost(a, b, c, XYZ):
    # squared second-difference in R^3 (curvature proxy)
    pa, pb, pc = XYZ[a], XYZ[b], XYZ[c]
    return float(np.linalg.norm(pc - 2*pb + pa)**2)

def _ring_dp_min_energy_cycle(G, XYZ, z_small, seed_loop, *,
                              r_hops=1,                # tighter search band
                              K_max=20,                # cap #candidates per position
                              w_data=1.0, w_len=0.25, w_smooth=1.4,
                              w_tv=0.4,               # NEW: HKS total-variation penalty
                              prefer_neck_mask=None,   # optional boolean mask to bias neck nodes
                              neck_bonus=0.3):         # subtract from z_small if node is neck
    """
    Dynamic programming on a ring (closed contour) with local candidate sets per position.
    Returns (best_cycle_nodes as np.ndarray) or None if no feasible cycle found.
    """
    m = int(len(seed_loop))
    if m < 3: return None

    # 1) Build candidate sets C[i] near each seed vertex (BFS out to r_hops)
    C = []
    for i, v in enumerate(seed_loop):
        fringe = {int(v)}
        visited = {int(v)}
        for _ in range(r_hops):
            newf = set()
            for u in fringe:
                for w in G.neighbors(u):
                    if w not in visited:
                        visited.add(int(w))
                        newf.add(int(w))
            fringe = newf

        cand = list(visited)
        # score candidates: low z_small and near the seed
        zs = z_small[cand].astype(float)
        if prefer_neck_mask is not None:
            zs = zs - neck_bonus * prefer_neck_mask[np.array(cand, int)]
        d0 = np.array([_euclid(v, u, XYZ) for u in cand], dtype=float)
        score = 1.0*zs + 0.05*d0
        order = np.argsort(score)
        sel = [cand[j] for j in order[:min(K_max, len(order))]]
        C.append(sel)

    if any(len(ci) == 0 for ci in C):
        return None

    best_total = np.inf
    best_seq = None

    data_costs = [np.array([float(z_small[u]) for u in Ci], dtype=float) for Ci in C]
    if prefer_neck_mask is not None:
        for i, Ci in enumerate(C):
            data_costs[i] = data_costs[i] - neck_bonus * prefer_neck_mask[np.array(Ci, int)]

    def feasible(u, v):
        return G.has_edge(int(u), int(v)) or (u == v)

    for s_idx, s in enumerate(C[0]):
        parents = []
        # i=0
        E0 = np.full((len(C[0]),), np.inf, dtype=float)
        E0[s_idx] = w_data * data_costs[0][s_idx]
        parents.append([(-1, -1)]*len(C[0]))
        E_prev = E0

        # i=1: length + data + TV to start anchor
        E1 = np.full((len(C[1]),), np.inf, dtype=float)
        par1 = [(-1, -1)]*len(C[1])
        for j, v1 in enumerate(C[1]):
            if not feasible(C[0][s_idx], v1):
                continue
            e_len = w_len * _euclid(C[0][s_idx], v1, XYZ)
            dz = abs(float(z_small[v1]) - float(z_small[C[0][s_idx]]))  # NEW TV term at i=1
            e = E_prev[s_idx] + e_len + w_data * data_costs[1][j] + w_tv * dz
            E1[j] = e
            par1[j] = (0, s_idx)
        parents.append(par1)
        E_prev = E1

        # i=2..m-1: length + smooth + data + TV
        for i in range(2, m):
            Ei = np.full((len(C[i]),), np.inf, dtype=float)
            pari = [(-1, -1)]*len(C[i])
            for j, vj in enumerate(C[i]):
                best_e = np.inf; best_k = -1
                for k, vk in enumerate(C[i-1]):
                    if not feasible(vk, vj):
                        continue
                    e_len = w_len * _euclid(vk, vj, XYZ)
                    pk, idx_pk = parents[i-1][k]
                    if pk == -1:
                        continue
                    vp = C[i-2][idx_pk]
                    e_smooth = w_smooth * _smooth_cost(vp, vk, vj, XYZ)
                    dz = abs(float(z_small[vj]) - float(z_small[vk]))  # NEW TV term
                    e = E_prev[k] + e_len + e_smooth + w_data * data_costs[i][j] + w_tv * dz
                    if e < best_e:
                        best_e = e; best_k = k
                if best_k != -1:
                    Ei[j] = best_e
                    pari[j] = (i-1, best_k)
            parents.append(pari)
            E_prev = Ei

        # closure (m-1)->0 with smooth + length + TV
        last_i = m-1
        best_e_total = np.inf
        best_last = -1
        for j, v_last in enumerate(C[last_i]):
            if not feasible(v_last, s):
                continue
            pk, idx_pk = parents[last_i][j]
            if pk == -1:
                continue
            v_prev = C[last_i-1][idx_pk]
            e_len_close = w_len * _euclid(v_last, s, XYZ)
            e_smooth_close = w_smooth * _smooth_cost(v_prev, v_last, s, XYZ)
            dz_close = abs(float(z_small[s]) - float(z_small[v_last]))  # NEW TV term at closure
            e_total = E_prev[j] + e_len_close + e_smooth_close + w_tv * dz_close
            if e_total < best_e_total:
                best_e_total = e_total
                best_last = j

        if best_last == -1:
            continue

        if best_e_total < best_total:
            seq_idx = [None]*m
            seq_idx[-1] = best_last
            for i in range(m-1, 0, -1):
                pi, pj = parents[i][seq_idx[i]]
                seq_idx[i-1] = pj
            seq_nodes = np.array([C[i][seq_idx[i]] for i in range(m)], dtype=int)
            best_total = best_e_total
            best_seq = seq_nodes

    return best_seq  # may be None


def _simple_cycle_from_selected(G, nodes_ordered):
    """
    Given an ordered selection (possibly with repeats), extract a simple cycle:
    - induce a subgraph on selected nodes + their 1-hop neighbors
    - take cycle_basis and pick the one with max overlap and largest length
    """
    sel = set(int(x) for x in nodes_ordered)
    cand = set(sel)
    for u in list(sel):
        cand.update(int(v) for v in G.neighbors(u))
    H = G.subgraph(list(cand)).copy()
    if H.number_of_nodes() < 3:
        return None
    cycles = nx.cycle_basis(H)
    if not cycles:
        return None
    Lset = sel
    best = max(cycles, key=lambda c: (len(set(c) & Lset), len(c)))
    return np.array(best, dtype=int)


# -------- Local snap & smooth (post-DP) --------

def _cycle_energy(loop, XYZ, z_small, w_data, w_len, w_smooth, w_tv):
    L = loop
    m = len(L)
    dist = 0.0; curv = 0.0; data = 0.0; tv = 0.0
    for i in range(m):
        a, b, c = L[(i-1)%m], L[i], L[(i+1)%m]
        dist += np.linalg.norm(XYZ[b]-XYZ[c])
        curv += np.linalg.norm(XYZ[c] - 2*XYZ[b] + XYZ[a])**2
        data += float(z_small[b])
        tv += abs(float(z_small[c]) - float(z_small[b]))
    return w_data*data + w_len*dist + w_smooth*curv + w_tv*tv

def _keeps_simple_cycle(G, loop):
    # ensure consecutive edges exist and all nodes unique (closure implicit)
    seen = set()
    m = len(loop)
    for i in range(m):
        u, v = int(loop[i]), int(loop[(i+1)%m])
        if not G.has_edge(u, v) and u != v:
            return False
        if i < m-1:
            if v in seen:
                return False
            seen.add(v)
    return True

def _separates_and_ok(G, loop, min_crypt_cells, crypt_pos_frac, z_small):
    loop_set = set(map(int, loop))
    Gc = G.copy(); Gc.remove_nodes_from(loop_set)
    touching = []
    for cc in nx.connected_components(Gc):
        cc = np.array(sorted(list(cc)), int)
        if any((nbr in loop_set) for n in cc for nbr in G.neighbors(int(n))):
            touching.append(cc)
    if len(touching) < 2: return False
    touching.sort(key=lambda a: a.size)
    crypt = touching[0]
    if crypt.size < min_crypt_cells: return False
    pos = np.mean((z_small[crypt] > 0).astype(float))
    return pos >= crypt_pos_frac

def local_snap_and_smooth(G, XYZ, z_small, loop, *,
                          w_data=1.0, w_len=0.25, w_smooth=1.4, w_tv=0.4,
                          max_iters=200,
                          min_crypt_cells=5, crypt_pos_frac=0.7):
    L = np.array(loop, int)
    bestE = _cycle_energy(L, XYZ, z_small, w_data, w_len, w_smooth, w_tv)

    def try_replace(idx, new_nodes):
        nonlocal L, bestE
        cand = L.copy()
        for t, nd in enumerate(new_nodes):
            cand[(idx+t)%len(cand)] = int(nd)
        if not _keeps_simple_cycle(G, cand):
            return False
        if not _separates_and_ok(G, cand, min_crypt_cells, crypt_pos_frac, z_small):
            return False
        E = _cycle_energy(cand, XYZ, z_small, w_data, w_len, w_smooth, w_tv)
        if E + 1e-9 < bestE:
            L, bestE = cand, E
            return True
        return False

    it = 0; improved = True
    while improved and it < max_iters:
        it += 1
        improved = False
        m = len(L)

        # 1) Shortcuts: u-v-w -> u-w
        for i in range(m):
            u, v, w = L[i-1], L[i], L[(i+1)%m]
            if G.has_edge(int(u), int(w)) and v != u and v != w:
                if try_replace(i, [w]):
                    improved = True
                    break
        if improved: continue

        # 2) 1-swap to smoother/lower-HKS neighbor
        for i in range(m):
            prev, cur, nxt = L[i-1], L[i], L[(i+1)%m]
            nbrs = [x for x in G.neighbors(int(cur))]
            nbrs = [x for x in nbrs if G.has_edge(int(prev), int(x)) and G.has_edge(int(x), int(nxt))]
            nbrs = sorted(nbrs, key=lambda x: (float(z_small[x]),
                                               np.linalg.norm(XYZ[nxt]-2*XYZ[int(x)]+XYZ[prev])**2))
            for vprime in nbrs[:4]:
                if vprime == cur:
                    continue
                if try_replace(i, [vprime]):
                    improved = True
                    break
            if improved: break
        if improved: continue

        # 3) Quad flip
        for i in range(m):
            u, v, w, x = L[i-1], L[i], L[(i+1)%m], L[(i+2)%m]
            if G.has_edge(int(u), int(w)) and G.has_edge(int(v), int(x)):
                if try_replace(i, [w]):
                    improved = True
                    break

    return L


def refine_loops_with_graph_snake(G, coords, z_small, seed_loops,
                                  *,
                                  r_hops=1, K_max=16,
                                  w_data=1.0, w_len=0.30, w_smooth=1.8, w_tv=0.5,
                                  neck_mask=None,
                                  min_crypt_cells=5,
                                  crypt_pos_frac=0.7,
                                  energy_tol=1e-9):
    """
    MONOTONE per-loop refinement:
      - local snap/smooth on the seed
      - optional DP proposal in a tight band
      - accept proposals ONLY if they reduce energy AND keep constraints
      - never drop loops; returns same number of loops as input
    """
    XYZ = coords[:, :3]
    refined = []

    # --- helpers reused from your code ---
    def separates(loop_poly):
        loop_set = set(int(x) for x in loop_poly)
        G_cut = G.copy(); G_cut.remove_nodes_from(loop_set)
        touching = []
        for cc in nx.connected_components(G_cut):
            cc = np.array(sorted(list(cc)), dtype=int)
            if any((nbr in loop_set) for n in cc for nbr in G.neighbors(int(n))):
                touching.append(cc)
        return (len(touching) >= 2), touching

    def crypt_ok(loop_poly):
        ok, touching = separates(loop_poly)
        if not ok: return False
        touching.sort(key=lambda a: a.size)
        crypt = touching[0]
        if crypt.size < min_crypt_cells: return False
        pos = np.mean((z_small[crypt] > 0).astype(float))
        return pos >= crypt_pos_frac

    def accept_if_better(curr_loop, cand_loop):
        """Return curr or cand depending on energy & constraints, guaranteeing no degradation."""
        if cand_loop is None:
            return curr_loop
        if not crypt_ok(cand_loop):
            return curr_loop
        Ec = _cycle_energy(curr_loop, XYZ, z_small, w_data, w_len, w_smooth, w_tv)
        En = _cycle_energy(cand_loop, XYZ, z_small, w_data, w_len, w_smooth, w_tv)
        if En + energy_tol < Ec:
            return cand_loop
        return curr_loop

    pref_mask = (neck_mask.astype(float) if neck_mask is not None else None)

    for seed in seed_loops:
        # 0) Ensure simple cycle baseline (don’t change order otherwise)
        base_try = _simple_cycle_from_selected(G, seed)
        if base_try is None:
            base = np.array(seed, dtype=int)
        else:
            base = base_try

        best = base.copy()  # always keep one loop

        # 1) LOCAL polish
        polished = local_snap_and_smooth(
            G, XYZ, z_small, best,
            w_data=w_data, w_len=w_len, w_smooth=w_smooth, w_tv=w_tv,
            max_iters=200,
            min_crypt_cells=min_crypt_cells, crypt_pos_frac=crypt_pos_frac
        )
        best = accept_if_better(best, polished)

        # 2) DP proposal in tight band
        dp = _ring_dp_min_energy_cycle(
            G, XYZ, z_small, best,
            r_hops=r_hops, K_max=K_max,
            w_data=w_data, w_len=w_len, w_smooth=w_smooth, w_tv=w_tv,
            prefer_neck_mask=pref_mask, neck_bonus=0.3
        )
        if dp is not None:
            dp_simpler = _simple_cycle_from_selected(G, dp)
            if dp_simpler is not None:
                dp = dp_simpler
        best = accept_if_better(best, dp)

        # 3) Final LIGHT local polish
        polished2 = local_snap_and_smooth(
            G, XYZ, z_small, best,
            w_data=w_data, w_len=w_len, w_smooth=w_smooth, w_tv=w_tv,
            max_iters=150,
            min_crypt_cells=min_crypt_cells, crypt_pos_frac=crypt_pos_frac
        )
        best = accept_if_better(best, polished2)

        refined.append(best)

    return refined




# --------------------------- MAIN PIPELINE ---------------------------

def dissect_by_necks(C: np.ndarray,
                     G: nx.Graph,
                     coords: np.ndarray,
                     *,
                     small_neg_thr: float = -0.5,   # neck: z_small <= this
                     large_near0_thr: float = 0.35, # neck: z_large <= this
                     min_loop_nodes: int = 8,       # discard tiny rings
                     band_hops: int = 1,            # neck band thinness (<= 1 hop off cycle)
                     min_crypt_cells: int = 5,      # interior region size guard
                     crypt_pos_frac: float = 0.7,   # interior positivity across all scales
                     do_plots: bool = True
                     ) -> list[nx.Graph]:
    """
    Clean neck-based dissection with minimal knobs.
    Steps:
      1) Neck candidates: small<=thr & |large|<=thr
      2) Extract loops per component (MCB) + split figure-eight (<=2 subloops)
      3) Thinness: keep loops whose band thickness <= band_hops
      4) Size: drop loops with < min_loop_nodes nodes
      5) Interior sanity: loop encloses crypt-like interior (size + positivity)
      6) Deduplicate: drop near-duplicate loops (Jaccard > 0.8, keep longer)
      7) Cut graph along union of loop nodes → return connected components
    """
    Z = C  # already z-scored
    z_small = Z[:, 0]
    z_large = Z[:, -1]

    def _print_and_plot(stage: str, loops: list[np.ndarray]):
        print(f"[{stage}] loops: {len(loops)}")
        if do_plots:
            fig = plot_organoid_loops_3d(coords, G, z_small, loops, title=f"{stage} — {len(loops)} loops")
            fig.show()

    # ------------ Step 1: neck candidates ------------
    neck_mask = (z_small <= small_neg_thr) & (z_large <= large_near0_thr)
    neck_sub = G.subgraph(np.where(neck_mask)[0]).copy()
    cand_components = [np.array(sorted(list(cc)), dtype=int) for cc in nx.connected_components(neck_sub)]
    draft_loops = []
    for cc in cand_components:
        H = neck_sub.subgraph(cc).copy()
        cycles = nx.cycle_basis(nx.k_core(H, 2)) if H.number_of_nodes() >= 3 else []
        if cycles:
            cyc = max(cycles, key=len)
            draft_loops.append(np.array(list(cyc), dtype=int))
    _print_and_plot("Step 1: candidates (draft cycles)", draft_loops)

    # ------------ Step 2: extract loops (MCB) & split figure-eight ------------
    def _thinness_ok_local(H_int: nx.Graph, loop_nodes_int: np.ndarray,
                        band_hops: int, max_outlier_frac: float = 0.05) -> bool:
        from collections import deque
        loop_set = set(int(x) for x in loop_nodes_int)
        dist = {n: np.inf for n in H_int.nodes()}
        dq = deque()
        for n in loop_set:
            dist[n] = 0
            dq.append(n)
        while dq:
            u = dq.popleft()
            if dist[u] >= band_hops:
                continue
            for v in H_int.neighbors(u):
                if dist[v] > dist[u] + 1:
                    dist[v] = dist[u] + 1
                    dq.append(v)
        band_nodes = [n for n, d in dist.items() if d <= band_hops]
        if not band_nodes:
            return False
        outliers = [n for n in band_nodes if dist[n] > band_hops]
        return (len(outliers) / max(1, len(band_nodes))) <= max_outlier_frac

    def _extract_loops_from_component(H: nx.Graph) -> list[np.ndarray]:
        nodes_orig = list(H.nodes())
        to_int = {n:i for i,n in enumerate(nodes_orig)}
        to_orig = {i:n for n,i in to_int.items()}
        H_int = nx.relabel_nodes(H, to_int, copy=True)
        mcb = nx.minimum_cycle_basis(H_int)
        if not mcb: return []
        mcb_sorted = sorted(mcb, key=len, reverse=True)

        def to_orig_cycle(c_int: list[int]) -> np.ndarray:
            return np.array([to_orig[int(u)] for u in c_int], dtype=int)

        loops = []
        main = np.array(mcb_sorted[0], dtype=int)
        if _thinness_ok_local(H_int, main, band_hops):
            loops.append(to_orig_cycle(main))
        if (H_int.number_of_edges() - H_int.number_of_nodes() + 1) >= 2 and len(mcb_sorted) > 1:
            a = set(main.tolist())
            for c in mcb_sorted[1:]:
                b = set(np.array(c, dtype=int).tolist())
                jacc = len(a & b) / max(1, len(a | b))
                if jacc > 0.8:
                    continue
                if _thinness_ok_local(H_int, np.array(c, dtype=int), band_hops):
                    loops.append(to_orig_cycle(c))
                    break
        return loops[:2]

    loops_step2: list[np.ndarray] = []
    for cc in cand_components:
        Hc = neck_sub.subgraph(cc).copy()
        if Hc.number_of_nodes() < 3:
            continue
        for block_nodes in nx.biconnected_components(Hc):
            B = Hc.subgraph(block_nodes).copy()
            if B.number_of_edges() - B.number_of_nodes() + 1 <= 0:
                continue
            loops_step2.extend(_extract_loops_from_component(B))
    _print_and_plot("Step 2: extracted loops (+ figure-eight split)", loops_step2)

    # ------------ Step 3 ------------
    loops_step3 = loops_step2
    _print_and_plot("Step 3: thinness check (band hops)", loops_step3)

    # ------------ Step 4 ------------
    loops_step4 = [lp for lp in loops_step3 if lp.size >= min_loop_nodes]
    _print_and_plot("Step 4: size filter", loops_step4)

    # ------------ Step 5: separation + shrink (as you had) ------------
    def _touching_components_for_loop(loop_poly: np.ndarray) -> list[np.ndarray]:
        loop_set = set(int(x) for x in loop_poly)
        G_cut = G.copy()
        G_cut.remove_nodes_from(loop_set)
        touching: list[np.ndarray] = []
        for cc in nx.connected_components(G_cut):
            cc_nodes = np.array(sorted(list(cc)), dtype=int)
            hit = False
            for n in cc_nodes:
                for nbr in G.neighbors(int(n)):
                    if nbr in loop_set:
                        hit = True; break
                if hit: break
            if hit:
                touching.append(cc_nodes)
        touching.sort(key=lambda a: a.size)
        return touching

    def _crypt_is_positive(crypt_nodes: np.ndarray) -> bool:
        pos_all = np.all(Z[crypt_nodes, :] > 0, axis=1)
        return (float(np.mean(pos_all)) >= crypt_pos_frac)

    def _order_loop_nodes(G: nx.Graph, loop_nodes: np.ndarray) -> np.ndarray:
        H = G.subgraph([int(x) for x in loop_nodes]).copy()
        if H.number_of_nodes() < 3:
            return np.array(loop_nodes, dtype=int)
        cycles = nx.cycle_basis(H)
        if not cycles:
            return np.array(loop_nodes, dtype=int)
        Lset = set(int(x) for x in loop_nodes)
        best = max(cycles, key=lambda c: (len(set(c) & Lset), len(c)))
        return np.array(best, dtype=int)

    def _ordered_cycle_including_node(G: nx.Graph, neck_mask: np.ndarray,
                                      loop_poly: np.ndarray, include_node: int) -> np.ndarray:
        cand = set(int(x) for x in loop_poly)
        for u in list(cand):
            for v in G.neighbors(u):
                if neck_mask[v]:
                    cand.add(int(v))
        cand.add(int(include_node))
        Hc = G.subgraph(list(cand)).copy()
        if Hc.number_of_nodes() < 3:
            return loop_poly
        cycles = nx.cycle_basis(Hc)
        if not cycles:
            return loop_poly
        Lset = set(int(x) for x in loop_poly)
        best = None
        best_key = (-1, -1)
        for c in cycles:
            if int(include_node) not in c:
                continue
            cset = set(int(x) for x in c)
            key = (len(cset & Lset), len(c))
            if key > best_key:
                best_key = key; best = c
        if best is None:
            return loop_poly
        return np.array(best, dtype=int)

    def _separates(G: nx.Graph, loop_poly: np.ndarray) -> bool:
        return (len(_touching_components_for_loop(loop_poly)) >= 2)

    neck_mask_for_step5 = (z_small <= small_neg_thr) & (z_large <= large_near0_thr)

    loops_step5: list[np.ndarray] = []
    for lp_idx, lp in enumerate(loops_step4):
        loop = np.array(lp, dtype=int)
        if not _separates(G, loop):
            print(f"[Step 5] loop#{lp_idx+1}: not separating; attempting to seal the loop (prefer neck, allow fallback)...")
            max_seal_iters = 200
            it = 0
            tried = set()
            while it < max_seal_iters and not _separates(G, loop):
                it += 1
                print(f"[Step 5][seal] loop#{lp_idx+1} it={it}: pre-add | loop len={loop.size}")
                if do_plots:
                    fig = plot_organoid_loops_3d(coords, G, z_large, [loop],
                                                 title=f"Step 5 • seal pre (loop#{lp_idx+1}, it={it})")
                    fig.show()
                loop_set = set(int(x) for x in loop)
                cand1_neck, cand1_fall = [], []
                for u in loop_set:
                    for v in G.neighbors(u):
                        if v in loop_set or v in tried:
                            continue
                        (cand1_neck if neck_mask_for_step5[v] else cand1_fall).append(int(v))
                cand2_neck, cand2_fall = [], []
                if not cand1_neck and not cand1_fall:
                    for u in loop_set:
                        for v in G.neighbors(u):
                            if v in loop_set:
                                continue
                            for w in G.neighbors(v):
                                if w in loop_set or w in tried:
                                    continue
                                (cand2_neck if neck_mask_for_step5[w] else cand2_fall).append(int(w))
                cand = []
                for group in (cand1_neck, cand1_fall, cand2_neck, cand2_fall):
                    if group:
                        group = np.array(sorted(set(group)), dtype=int)
                        group = group[np.argsort(z_small[group])]
                        cand.extend(group.tolist())
                print(f"[Step 5][seal] loop#{lp_idx+1} it={it}: candidates: "
                      f"1hop-neck={len(cand1_neck)}, 1hop-fallback={len(cand1_fall)}, "
                      f"2hop-neck={len(cand2_neck)}, 2hop-fallback={len(cand2_fall)}")
                if not cand:
                    print(f"[Step 5][seal] loop#{lp_idx+1} it={it}: no candidates left (even with fallback).")
                    break
                progressed = False
                for add_node in cand:
                    tried.add(add_node)
                    new_loop = _ordered_cycle_including_node(G, neck_mask_for_step5, loop, add_node)
                    if set(new_loop.tolist()) == set(loop.tolist()):
                        continue
                    if new_loop.size < int(0.8 * loop.size):
                        print(f"[Step 5][seal] loop#{lp_idx+1} it={it}: REJECT new cycle len={new_loop.size} "
                              f"(shrinks from {loop.size})."); continue
                    overlap = len(set(new_loop.tolist()) & set(loop.tolist()))
                    if overlap < int(0.5 * loop.size):
                        print(f"[Step 5][seal] loop#{lp_idx+1} it={it}: REJECT new cycle overlap={overlap} "
                              f"(old len={loop.size})."); continue
                    loop = new_loop
                    print(f"[Step 5][seal] loop#{lp_idx+1} it={it}: added node {add_node} "
                          f"(neck={bool(neck_mask_for_step5[add_node])}, z_small={z_small[add_node]:.3f}); "
                          f"new loop len={loop.size}")
                    if do_plots:
                        fig = plot_organoid_loops_3d(coords, G, z_large, [loop],
                                                     title=f"Step 5 • seal post-add {add_node} "
                                                           f"(loop#{lp_idx+1}, it={it})")
                        fig.show()
                    progressed = True
                    if _separates(G, loop):
                        print(f"[Step 5][seal] loop#{lp_idx+1} it={it}: loop now separates.")
                        break
                if not progressed:
                    print(f"[Step 5][seal] loop#{lp_idx+1} it={it}: made no progress (cycle couldn’t absorb candidates).")
                    break
            if not _separates(G, loop):
                print(f"[Step 5] DROP loop#{lp_idx+1}: could not seal to a separating cycle (even with fallback).")
                continue
            else:
                print(f"[Step 5] loop#{lp_idx+1}: sealed successfully.")

        touching = _touching_components_for_loop(loop)
        if len(touching) == 0:
            print(f"[Step 5] DROP loop#{lp_idx+1}: no adjacent components after cut.")
            continue
        crypt = touching[0]
        villus = touching[-1]
        print(f"[Step 5] loop#{lp_idx+1}: initial crypt size={crypt.size}, villus size={villus.size}, loop len={loop.size}")

        if crypt.size < min_crypt_cells:
            print(f"[Step 5] DROP loop#{lp_idx+1}: crypt too small ({crypt.size} < {min_crypt_cells}).")
            continue
        if loop.size < min_loop_nodes:
            print(f"[Step 5] DROP loop#{lp_idx+1}: loop too small ({loop.size} < {min_loop_nodes}).")
            continue

        if _crypt_is_positive(crypt):
            print(f"[Step 5] KEEP loop#{lp_idx+1}: crypt positive without shrink.")
            loops_step5.append(loop)
            continue

        max_iters = 200
        it = 0
        rejected_reason = None
        while True:
            it += 1
            if it > max_iters:
                rejected_reason = f"max iters exceeded ({max_iters}) while shrinking"
                crypt = np.array([], dtype=int)
                break
            loop_set = set(int(x) for x in loop)
            boundary = []
            for n in crypt:
                if any((nbr in loop_set) for nbr in G.neighbors(int(n))):
                    boundary.append(int(n))
            if not boundary:
                rejected_reason = "no boundary crypt nodes left to shrink"
                crypt = np.array([], dtype=int)
                break
            b = np.array(boundary, dtype=int)
            move = int(b[np.argmin(z_small[b])])
            print(f"[Step 5] loop#{lp_idx+1} it={it}: move boundary node {move} (z_small={z_small[move]:.3f}) into loop")
            loop = _ordered_cycle_including_node(G, neck_mask_for_step5, loop, move)
            if not _separates(G, loop):
                rejected_reason = "loop lost separation after shrink step"
                crypt = np.array([], dtype=int)
                break
            touching = _touching_components_for_loop(loop)
            if len(touching) == 0:
                rejected_reason = "degenerate cut after shrink (no touching components)"
                crypt = np.array([], dtype=int)
                break
            crypt = touching[0]; villus = touching[-1]
            print(f"[Step 5] loop#{lp_idx+1} it={it}: new crypt size={crypt.size}, loop len={loop.size}")
            if crypt.size < min_crypt_cells:
                rejected_reason = f"crypt too small after shrink ({crypt.size} < {min_crypt_cells})"
                crypt = np.array([], dtype=int); break
            if loop.size < min_loop_nodes:
                rejected_reason = f"loop too small after shrink ({loop.size} < {min_loop_nodes})"
                crypt = np.array([], dtype=int); break
            if _crypt_is_positive(crypt):
                print(f"[Step 5] KEEP loop#{lp_idx+1}: crypt positive after shrink (iters={it}, crypt size={crypt.size}).")
                break

        if crypt.size == 0:
            print(f"[Step 5] DROP loop#{lp_idx+1}: {rejected_reason}.")
            continue

        loops_step5.append(loop)

    _print_and_plot("Step 5: separation+assign-to-villus+shrink (order preserved, sealing if needed)", loops_step5)

    # ------------ Step 6: deduplicate ------------
    def _dedup(loops: list[np.ndarray]) -> list[np.ndarray]:
        keep: list[np.ndarray] = []
        for lp in sorted(loops, key=lambda a: -a.size):
            aset = set(lp.tolist())
            dup = False
            for kp in keep:
                bset = set(kp.tolist())
                jacc = len(aset & bset) / max(1, len(aset | bset))
                if jacc > 0.8:
                    dup = True; break
            if not dup:
                keep.append(lp)
        return keep
    loops_final = _dedup(loops_step5)
    _print_and_plot("Step 6: deduplicate", loops_final)

    # ------------ Step 7: cut graph along loop nodes ------------
    if len(loops_final) == 0:
        _print_and_plot("Step 7: no loops — return whole graph", [])
        return [G.copy()]
    cut_nodes = np.unique(np.concatenate(loops_final))
    G_cut = G.copy(); G_cut.remove_nodes_from(cut_nodes)
    pieces = [G.subgraph(cc).copy() for cc in nx.connected_components(G_cut)]
    _print_and_plot("Step 7: cut & components", loops_final)
    return pieces


# --------------------------- RUN ---------------------------

HKS_zscored = (cell_weighted_hks-np.mean(cell_weighted_hks, axis=0))/np.std(cell_weighted_hks, axis=0)

components = dissect_by_necks(
    HKS_zscored[:,:7], cell_graph, centroids,
    small_neg_thr= -0.1,
    large_near0_thr=0.4,
    min_loop_nodes=8,
    band_hops=2,
    min_crypt_cells=5,
    crypt_pos_frac=0.5,
    do_plots=True
)
print("num components:", len(components))
```

```python
from __future__ import annotations
import numpy as np
import networkx as nx
import plotly.graph_objects as go

# ---------------------------
# Step 1 — low-HKS regions
# ---------------------------

def make_low_mask(HKS, thr_t0, thr_all):
    """
    Binary mask of nodes that satisfy BOTH:
      - HKS at the smallest time-scale (col 0) <= thr_t0
      - HKS at ALL time-scales <= thr_all
    """
    if HKS.ndim != 2 or HKS.shape[1] < 1:
        raise ValueError("HKS must be a 2D array with at least 1 time-scale (column).")
    mask_t0 = HKS[:, 0] <= thr_t0
    mask_all = np.all(HKS <= thr_all, axis=1)
    return mask_t0 & mask_all


def find_low_regions(G, low_mask, min_region_size=5):
    """
    Find connected components of the subgraph induced by low_mask.
    Keep only components with at least 'min_region_size' nodes.
    """
    if G.number_of_nodes() != low_mask.shape[0]:
        raise ValueError("G and low_mask size mismatch.")
    nodes_low = [n for n in G.nodes if low_mask[n]]
    sub = G.subgraph(nodes_low)
    comps = [set(c) for c in nx.connected_components(sub)]
    comps = [c for c in comps if len(c) >= min_region_size]
    return comps


# ---------------------------
# Step 2 — loops on region boundary
# ---------------------------

def region_boundary_nodes(G, region):
    """
    Nodes of the region that touch at least one neighbor outside the region.
    These approximate the geometric boundary of the low-HKS patch.
    """
    boundary = set()
    region_set = set(region)
    for u in region:
        for v in G.neighbors(u):
            if v not in region_set:
                boundary.add(u)
                break
    return boundary


def boundary_subgraph(G, boundary_nodes):
    """
    Subgraph induced by the boundary nodes (edges only between boundary nodes).
    """
    return G.subgraph(boundary_nodes).copy()


def extract_loops_on_boundary(B, min_loop_len=5):
    """
    Use an undirected cycle basis on the boundary subgraph.
    Keep cycles of length >= min_loop_len.
    Returns each loop as a numpy array of node indices.
    """
    if B.number_of_nodes() == 0:
        return []
    loops = nx.cycle_basis(B)  # list of lists of node ids
    loops = [np.asarray(c, dtype=int) for c in loops if len(c) >= min_loop_len]
    return loops


# ---------------------------
# Debug visualization helpers
# ---------------------------

def add_region_overlays(fig, coords, regions):
    """
    Overlay translucent colored markers for each Step-1 region.
    Keeps the base figure intact and just adds more traces.
    """
    XYZ = coords[:, :3]
    region_colors = [
        "rgba(227, 26, 28, 0.25)",  # red
        "rgba(31, 120, 180, 0.25)", # blue
        "rgba(51, 160, 44, 0.25)",  # green
        "rgba(106, 61, 154, 0.25)", # purple
        "rgba(255, 127, 0, 0.25)",  # orange
        "rgba(166, 86, 40, 0.25)",  # brown
        "rgba(152, 78, 163, 0.25)", # violet
        "rgba(0, 0, 0, 0.25)",      # gray
    ]
    for k, reg in enumerate(regions):
        idx = np.fromiter(reg, dtype=int)
        if idx.size == 0:
            continue
        fig.add_trace(go.Scatter3d(
            x=XYZ[idx, 0], y=XYZ[idx, 1], z=XYZ[idx, 2],
            mode="markers",
            marker=dict(size=6.5, color=region_colors[k % len(region_colors)]),
            name=f"region {k+1}",
            showlegend=True,
            hoverinfo="skip",
        ))


# ---------------------------
# Orchestrator
# ---------------------------

def detect_neck_loops(
    coords,
    G,
    HKS,
    thr_t0,
    thr_all,
    min_region_size=5,
    min_loop_len=5,
    plot_func=None,
    title="Neck detection (debug)"
):
    """
    Run Step-1 (low regions) and Step-2 (boundary loops) with simple parameters.

    Thresholds:
      - thr_t0: threshold for HKS[:,0]
      - thr_all: threshold to hold for all time-scales

    Cutoffs:
      - min_region_size: drop Step-1 components smaller than this
      - min_loop_len: drop Step-2 loops shorter than this

    If 'plot_func' is provided (e.g., plot_organoid_loops_3d), a 3D debug figure is created:
      * nodes colored by z_small = HKS[:,0]
      * extracted loops rendered as thick polylines
      * Step-1 regions overlaid as translucent marker clouds
    """
    # Step 1: thresholding and regions
    low_mask = make_low_mask(HKS, thr_t0=thr_t0, thr_all=thr_all)
    regions = find_low_regions(G, low_mask, min_region_size=min_region_size)

    # Step 2: loops on region boundaries
    loops_by_region = []
    for k, reg in enumerate(regions):
        bnodes = region_boundary_nodes(G, reg)
        B = boundary_subgraph(G, bnodes)
        loops_k = extract_loops_on_boundary(B, min_loop_len=min_loop_len)
        loops_by_region.append(loops_k)
        print(f"Region {k+1}: {len(loops_k)} loop(s)")

    # Optional plotting
    fig = None
    if plot_func is not None:
        all_loops = [loop for loops_k in loops_by_region for loop in loops_k]
        z_small = HKS[:, 0]
        fig = plot_func(coords=coords, G=G, z_small=z_small, loops=all_loops, title=title)
        add_region_overlays(fig, coords, regions)

    return {
        "low_mask": low_mask,
        "regions": regions,
        "loops_by_region": loops_by_region,
        "figure": fig
    }


HKS_zscored = (cell_weighted_hks-np.mean(cell_weighted_hks, axis=0))/np.std(cell_weighted_hks, axis=0)

# Example call (assuming you defined plot_organoid_loops_3d from your snippet):
result = detect_neck_loops(
    coords=centroids,         # (N,3)
    G=cell_graph,                   # networkx graph with N nodes labeled 0..N-1
    HKS=HKS_zscored[:,:5],               # (N,T)
    thr_t0=-0.1,          # threshold for the smallest time-scale (col 0)
    thr_all=0.2,          # threshold that must hold for ALL time-scales
    min_region_size=5,     # per your constraint (B)
    min_loop_len=5,        # per your constraint (B)
    plot_func=plot_organoid_loops_3d,
    title="Organoid necks (debug)"
)

# Access:
fig = result["figure"]
if fig is not None:
    fig.show()
```

```python
from __future__ import annotations
import numpy as np
import networkx as nx
import plotly.graph_objects as go

# ---------------------------
# Step 1 — high-HKS mask (all time-scales)
# ---------------------------

def make_high_mask(HKS, thr_allpos):
    """
    Binary mask of nodes that satisfy:
      - HKS at ALL time-scales (columns) >= thr_allpos
    """
    if HKS.ndim != 2 or HKS.shape[1] < 1:
        raise ValueError("HKS must be a 2D array with at least 1 time-scale (column).")
    return np.all(HKS >= thr_allpos, axis=1)


def find_high_regions(G, high_mask, min_region_size=5):
    """
    Find connected components of the subgraph induced by 'high_mask'.
    Keep only components with at least 'min_region_size' nodes.
    Each returned region is a set of node indices (connected patch).
    """
    if G.number_of_nodes() != high_mask.shape[0]:
        raise ValueError("G and high_mask size mismatch.")
    nodes_high = [n for n in G.nodes if high_mask[n]]
    sub = G.subgraph(nodes_high)
    comps = [set(c) for c in nx.connected_components(sub)]
    comps = [c for c in comps if len(c) >= min_region_size]
    return comps


# ---------------------------
# Debug visualization helper (overlays for crypt regions)
# ---------------------------

def add_crypt_overlays(fig, coords, regions):
    """
    Overlay translucent colored markers for each crypt region.
    """
    XYZ = coords[:, :3]
    region_colors = [
        "rgba(227, 26, 28, 0.25)",  # red
        "rgba(31, 120, 180, 0.25)", # blue
        "rgba(51, 160, 44, 0.25)",  # green
        "rgba(106, 61, 154, 0.25)", # purple
        "rgba(255, 127, 0, 0.25)",  # orange
        "rgba(166, 86, 40, 0.25)",  # brown
        "rgba(152, 78, 163, 0.25)", # violet
        "rgba(0, 0, 0, 0.25)",      # gray
    ]
    for k, reg in enumerate(regions):
        idx = np.fromiter(reg, dtype=int)
        if idx.size == 0:
            continue
        fig.add_trace(go.Scatter3d(
            x=XYZ[idx, 0], y=XYZ[idx, 1], z=XYZ[idx, 2],
            mode="markers",
            marker=dict(size=6.5, color=region_colors[k % len(region_colors)]),
            name=f"crypt {k+1}",
            showlegend=True,
            hoverinfo="skip",
        ))


import numpy as np
import networkx as nx

def _false_components(G, high_mask):
    """Connected components induced by nodes where high_mask == False."""
    false_nodes = [n for n in G.nodes if not high_mask[n]]
    sub = G.subgraph(false_nodes)
    return [set(c) for c in nx.connected_components(sub)]

def _is_enclosed(G, comp, high_mask):
    """Check if every neighbor just outside 'comp' is True in high_mask."""
    boundary_neighbors = set()
    for u in comp:
        for v in G.neighbors(u):
            if v not in comp:
                boundary_neighbors.add(v)
    if not boundary_neighbors:
        # Isolated pocket with no outside neighbors (unlikely); treat as enclosed.
        return True
    # Enclosed iff all boundary neighbors are True
    return all(high_mask[v] for v in boundary_neighbors)

def fill_small_crypt_defects(G, high_mask, max_hole_size=5, verbose=True):
    """
    Fill small 'defect' patches (False) that are completely surrounded by True nodes.
    A defect is filled if:
      - it is a connected component of False nodes of size <= max_hole_size, and
      - ALL its boundary neighbors (outside the component) are True.

    Parameters
    ----------
    G : networkx.Graph
        Graph with N nodes labeled 0..N-1.
    high_mask : np.ndarray (bool, shape (N,))
        True for nodes that meet the crypt threshold across all time-scales.
    max_hole_size : int
        Maximum size of a defect component to fill (default 5).
    verbose : bool
        Print how many patches were filled.

    Returns
    -------
    high_mask_filled : np.ndarray (bool, shape (N,))
        Updated mask with small enclosed defects flipped to True.
    filled_patches : list of sets
        Each set is the node indices of a filled defect component.
    """
    if high_mask.dtype != bool:
        raise ValueError("high_mask must be a boolean array.")
    if G.number_of_nodes() != high_mask.shape[0]:
        raise ValueError("G and high_mask size mismatch.")

    high_mask_filled = high_mask.copy()
    filled_patches = []

    comps = _false_components(G, high_mask_filled)
    for comp in comps:
        if len(comp) <= max_hole_size and _is_enclosed(G, comp, high_mask_filled):
            # Fill this defect: flip to True
            idx = np.fromiter(comp, dtype=int)
            high_mask_filled[idx] = True
            filled_patches.append(comp)

    if verbose:
        print(f"Filled {len(filled_patches)} enclosed defect patch(es) (size ≤ {max_hole_size}).")

    return high_mask_filled, filled_patches


import numpy as np
import networkx as nx

# ---------------------------
# Boundary & loop extraction
# ---------------------------

def region_boundary_nodes(G, region):
    """Nodes of 'region' that touch at least one neighbor outside the region."""
    boundary = set()
    region_set = set(region)
    for u in region:
        for v in G.neighbors(u):
            if v not in region_set:
                boundary.add(u)
                break
    return boundary

def boundary_subgraph(G, boundary_nodes):
    """Subgraph induced by boundary nodes."""
    return G.subgraph(boundary_nodes).copy()

def extract_loops_on_boundary(B, min_loop_len=5):
    """
    Use cycle basis on the boundary subgraph.
    Keep cycles of length >= min_loop_len.
    """
    if B.number_of_nodes() == 0:
        return []
    loops = nx.cycle_basis(B)
    loops = [np.asarray(c, dtype=int) for c in loops if len(c) >= min_loop_len]
    return loops

def initial_boundary_loops_for_region(G, region, min_loop_len=5):
    """Convenience: get initial boundary loops for one region."""
    bnodes = region_boundary_nodes(G, region)
    B = boundary_subgraph(G, bnodes)
    return extract_loops_on_boundary(B, min_loop_len=min_loop_len)

# ---------------------------
# Loop utilities
# ---------------------------

def loop_cost(z_small, loop_nodes):
    """Sum of z_small along the loop (lower is better)."""
    return float(np.sum(z_small[np.asarray(loop_nodes, dtype=int)]))

def _neighbors_that_link_prev_next(G, prev_n, next_n):
    """
    Return neighbors 'w' that are adjacent to BOTH prev_n and next_n
    (i.e., allow replacing the middle node with 'w' while keeping a simple cycle).
    Includes possibility that prev_n and/or next_n themselves appear, but those
    will be filtered by caller.
    """
    # intersect neighbor sets
    n_prev = set(G.neighbors(prev_n))
    n_next = set(G.neighbors(next_n))
    return n_prev & n_next

def _tighten_once(G, loop_nodes, z_small, forbidden, allow_remove=True):
    """
    One greedy coordinate-descent step over the loop.
    Tries either:
      - removal of a node v if prev-next edge exists (shrinks length), or
      - replacement of v by a neighbor w that connects to both prev and next and lowers cost.
    Constraints:
      - 'forbidden' nodes (crypt region) cannot be part of the loop.
      - Loop length must not increase (removal decreases; replacement keeps same).
      - Replacement 'w' must not already be in the loop (except prev/next).
    Returns (new_loop_nodes, improved: bool).
    """
    L = len(loop_nodes)
    if L < 3:
        return loop_nodes, False

    loop = list(loop_nodes)
    loop_set = set(loop)

    best_delta = 0.0
    best_action = None  # ("remove", idx) or ("replace", idx, w)
    current_cost = loop_cost(z_small, loop)

    for i in range(L):
        prev_i = (i - 1) % L
        next_i = (i + 1) % L
        v = loop[i]
        p = loop[prev_i]
        n = loop[next_i]

        # --- Option A: removal (shrink) if prev and next are adjacent
        if allow_remove and G.has_edge(p, n):
            # New loop would be loop without v
            # Ensure we don't bring in forbidden nodes (we don't add any node).
            # Cost delta: - z_small[v]
            delta = -float(z_small[v])
            if delta < best_delta:  # we minimize cost (delta negative is better)
                best_delta = delta
                best_action = ("remove", i)

        # --- Option B: replacement (slide) to a neighbor w that links p and n
        candidates = _neighbors_that_link_prev_next(G, p, n)
        for w in candidates:
            if w == v or w == p or w == n:
                continue
            if w in forbidden:
                continue
            if w in loop_set:
                continue  # avoid self-intersections / reuse
            # Cost delta if we replace v with w (same length)
            delta = float(z_small[w] - z_small[v])
            if delta < best_delta:
                best_delta = delta
                best_action = ("replace", i, w)

    if best_action is None:
        return loop_nodes, False

    # Apply the best local action
    if best_action[0] == "remove":
        i = best_action[1]
        new_loop = loop[:i] + loop[i+1:]
        return np.asarray(new_loop, dtype=int), True
    else:
        _, i, w = best_action
        new_loop = loop.copy()
        new_loop[i] = w
        return np.asarray(new_loop, dtype=int), True

def tighten_loop(G, loop_nodes, z_small, forbidden, min_loop_len=5, max_iters=200):
    """
    Iteratively tighten a loop by greedy local moves that never increase length and
    (ideally) lower the total z_small cost. Stops when no improvement occurs.
    - 'forbidden' are crypt nodes: loop may not enter them.
    - If loop shrinks below 'min_loop_len', it is discarded (returns None).
    """
    loop = np.asarray(loop_nodes, dtype=int)
    if loop.size < min_loop_len:
        return None

    prev_cost = loop_cost(z_small, loop)
    for _ in range(max_iters):
        loop_new, improved = _tighten_once(G, loop, z_small, forbidden, allow_remove=True)
        if not improved:
            break
        if loop_new.size < min_loop_len:
            return None
        # If length same but cost increased (shouldn’t happen with our selection), guard:
        new_cost = loop_cost(z_small, loop_new)
        if loop_new.size > loop.size:
            # Should never happen with our moves, but enforce non-increase:
            break
        loop = loop_new
        prev_cost = new_cost

    # Final pass without allowing removals, to settle geometrically without over-shrinking
    for _ in range(50):
        loop_new, improved = _tighten_once(G, loop, z_small, forbidden, allow_remove=False)
        if not improved:
            break
        if loop_new.size < min_loop_len:
            return None
        loop = loop_new

    return np.asarray(loop, dtype=int)

# ---------------------------
# Orchestrator (updated)
# ---------------------------

def detect_crypt_patches(
    coords,
    G,
    HKS,
    thr_allpos,
    min_region_size=5,
    plot_func=None,
    title="Crypt detection (debug)"
):
    """
    Identify crypt-like patches (HKS >= thr_allpos at all time-scales), fill small defects,
    then extract boundary loops and tighten them away from the crypt to lower HKS (t=small),
    never increasing loop length. Plots crypt regions and final loops.
    """
    # Step 1: thresholding across all time-scales
    high_mask = make_high_mask(HKS, thr_allpos=thr_allpos)

    # Fill tiny enclosed defects (≤5 nodes)
    high_mask_filled, _ = fill_small_crypt_defects(G, high_mask, max_hole_size=10)

    # Step 2: connected crypt regions
    regions = find_high_regions(G, high_mask_filled, min_region_size=min_region_size)
    print(f"Found {len(regions)} crypt region(s).")
    for i, reg in enumerate(regions, 1):
        print(f"  - Crypt {i}: {len(reg)} nodes")

    # ---------------------------
    # Step 3: initial boundary loops per region
    # ---------------------------
    initial_loops_by_region = []
    for reg in regions:
        loops0 = initial_boundary_loops_for_region(G, reg, min_loop_len=5)
        initial_loops_by_region.append(loops0)

    initial_loops = [lp for loops0 in initial_loops_by_region for lp in loops0]


    # ---------------------------
    # Step 4: tighten loops away from crypt (toward lower HKS at small t)
    # ---------------------------
    z_small = HKS[:, 0]
    final_loops = []
    for reg, loops0 in zip(regions, initial_loops_by_region):
        forb = set(reg)  # crypt cells are forbidden for the neck loop
        for loop0 in loops0:
            loop_final = tighten_loop(G, loop0, z_small, forbidden=forb, min_loop_len=5, max_iters=200)
            if loop_final is not None:
                final_loops.append(loop_final)

    print(f"Tightened to {len(final_loops)} loop(s) total.")

    # Optional plotting
    fig = None
    if plot_func is not None:
        fig = plot_func(coords=coords, G=G, z_small=z_small, loops=initial_loops, title=title)
        add_crypt_overlays(fig, coords, regions)

    return {
        "high_mask": high_mask,
        "regions": regions,
        "loops": final_loops,
        "figure": fig
    }



HKS_zscored = (cell_weighted_hks-np.mean(cell_weighted_hks, axis=0))/np.std(cell_weighted_hks, axis=0)


result = detect_crypt_patches(
    coords=centroids,
    G=cell_graph,
    HKS=HKS_zscored[:,:6],
    thr_allpos=0.4,     # example: require HKS >= 0.15 for every time-scale
    min_region_size=5,
    plot_func=plot_organoid_loops_3d,
    title="Crypt patches (debug)"
)

fig = result["figure"]
if fig is not None:
    fig.show()



```
